```python
import pandas as pd
import numpy as np

df = pd.read_csv("data.csv")
df.shape
```

```python
X = df.drop(["id", "diagnosis"], axis=1)
y = df["diagnosis"]

print(X.shape, y.shape)
```

```python
y.replace("M", 1, inplace=True)
y.replace("B", 0, inplace=True)
```

```python
def eucledian_distance(person1, person2):
    return np.sqrt(np.square(person2 - person1).sum())
```

```python
def get_distances(new_person):
    distances = {}
    for index, person in X.iterrows():
        distances[index] = eucledian_distance(person, new_person)

    distances = sorted(distances, key=distances.get)
    return distances
```

```python
def knn(new_person, k=5):
    distances = get_distances(new_person)
    top_labels = []
    for index in distances[:k]:
        top_labels.append(y[index])

    top_labels = pd.Series(top_labels)

    return top_labels.mode()[0]
```

```python
predictions = []
for index, person in X.iterrows():
    predictions.append(knn(person))

predictions
```

```python
from sklearn.metrics import classification_report

classification_report(y.astype(int), predictions)
```
